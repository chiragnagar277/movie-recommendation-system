```python
movie_ds = ds.groupby("Movie_Title").agg({
    "Genre": lambda x: x.mode()[0] if not x.mode().empty else "Unknown",
    "Release_Year": "mean",
    "Rating": "mean",
    "Number_of_Ratings": "mean",
    "Runtime_Minutes": "mean",
    "Language": lambda x: x.mode()[0] if not x.mode().empty else "Unknown",
    "Age_Group": lambda x: x.mode()[0] if not x.mode().empty else "Unknown"
}).reset_index()
```

```python
features = [
    "Genre",
    "Release_Year",
    "Rating",
    "Number_of_Ratings",
    "Runtime_Minutes",
    "Language",
    "Age_Group"
]

X = movie_ds[features]
```

```python
X_encoded = pd.get_dummies(
    X,
    columns=["Genre", "Language", "Age_Group"],
    dtype=int
)
```

```python
print(X_encoded.shape)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X_encoded)
```

```python
from sklearn.cluster import KMeans
```

```python
inertia = []
for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    
    inertia.append(kmeans.inertia_)
```

```python
for k, value in zip(range(2, 11), inertia):
    print(f"K = {k}, Inertia = {value:.2f}")
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    inertia,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.show()
```

```python
optimal_k = 5

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

movie_ds["Cluster"] = kmeans.fit_predict(X_scaled)
```

```python
def recommend_movies(movie_title, n=5):
    movie = movie_ds[
        movie_ds["Movie_Title"].str.strip().str.lower()
        == movie_title.strip().lower()
    ]
    if movie.empty:
        print("Movie not found.")
        return pd.DataFrame()

    movie_index = movie.index[0]
    selected_cluster = movie_ds.loc[
        movie_index, "Cluster"
    ]

    # Get similarity scores
    similarity_scores = cosine_sim[movie_index]
    candidates = []

    for idx in movie_ds.index:
        if idx == movie_index:
            continue

        candidates.append({
            "index": idx,
            "Movie_Title": movie_ds.loc[idx, "Movie_Title"],
            "Genre": movie_ds.loc[idx, "Genre"],
            "Rating": movie_ds.loc[idx, "Rating"],
            "Cluster": movie_ds.loc[idx, "Cluster"],
            "Similarity": similarity_scores[idx]
        })

    # Same-cluster movies first,
    # then sort by similarity
    candidates.sort(
        key=lambda x: (
            x["Cluster"] != selected_cluster,
            -x["Similarity"]
        )
    )

    # Take top N
    recommendations = candidates[:n]

    # Convert to DataFrame
    result = pd.DataFrame(recommendations)

    # Round similarity
    result["Similarity"] = result["Similarity"].round(3)

    return result[
        [
            "Movie_Title",
            "Genre",
            "Rating",
            "Similarity"
        ]
    ]
```

```python
recommend_movies("Future World")
```

```python
import joblib

model = joblib.load("model/recommender.pkl")

print(model.keys())
```

```python
model["movie_ds"]["Movie_Title"].head(20)
```

```python
def recommend_movies(movie_name, n=5):

    movie_ds = model["movie_ds"]
    cosine_sim = model["cosine_sim"]

    # Find the movie
    matches = movie_ds[
        movie_ds["Movie_Title"].str.lower() == movie_name.lower()
    ]

    if matches.empty:
        print("Movie not found!")
        return

    # Get position of movie
    index = movie_ds.index.get_loc(matches.index[0])

    # Get similarity scores
    scores = list(enumerate(cosine_sim[index]))

    # Sort from highest to lowest
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    print("Recommended movies for:", movie_name)
    print()

    # Display recommendations
    for i, score in scores[1:n+1]:
        print(movie_ds.iloc[i]["Movie_Title"])
```
